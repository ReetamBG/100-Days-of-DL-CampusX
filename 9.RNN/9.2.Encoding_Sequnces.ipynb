```python
data = [
    "Hello, my name is Ree",
    "Hi, my name is Bee",
    "What are you doing ?",
    "The movie was good",
    "The movie was not that good",
    "It was bad"
]
```

### Integer Encoding

```python
from tensorflow.keras.preprocessing.text import Tokenizer

tokenizer = Tokenizer(oov_token="<unkown word>")        # replace with oov_token when unseen data appears
tokenizer.fit_on_texts(data)
```

```python
tokenizer.word_index        # get the index(integer) assinged to each unique word
tokenizer.word_counts       # occurence of each word in the entire data
tokenizer.document_count    # the number of samples in the document(data)

len(tokenizer.word_index)       # to get all unique words
```

```python
# to create the sequences

sequences = tokenizer.texts_to_sequences(["Hello, how are you"])
sequences
```

```python
sequences = tokenizer.texts_to_sequences(data)
sequences
```

```python
# To add padding

from tensorflow.keras.utils import pad_sequences

pad_sequences(sequences, padding="post")        # post padding -> zeros at the last
```
